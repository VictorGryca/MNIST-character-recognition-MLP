# Testando lógica treinando uma rede XOR

```python
# importando as bibliotecas essenciais
import numpy as np 
```

```python
# datasets XOR e pesos iniciais
np.random.seed(42)

X = np.array([[1,1],[1,0],[0,1],[0,0]])  # (4, 2)
y = np.array([[0],[1],[1],[0]])          # (4, 1)

W1 = np.random.uniform(-0.1, 0.1, (4, 2))  # (4 neurônios, 2 entradas) | quantidade no feeling de neuronios, acho que dava pra fazer com 2
b1 = np.zeros((4,))
W2 = np.random.uniform(-0.1, 0.1, (1, 4))  # (1 saída, 4 neurônios)
b2 = np.zeros((1,))

Lr = 0.2
```

```python
# @ — multiplicação matricial (W @ x)
# * — multiplicação elemento a elemento
# np.outer() — produto externo (só no backprop)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x)) # 1 é porcentagem mais alta e 0 é mais baixa | funcao ativacao binaria
```

```python
def derivada_relu(z):
    return (z > 0).astype(float)  # 1 se z era positivo (neuronio ligado), 0 se nao era
```

```python
for epoca in range(1000):
    for i in range(len(X)):
        # forward pass: passa o exemplo pela rede
        x = X[i]           # pega um exemplo por vez
        z1 = W1 @ x + b1  # combinacao linear da camada 1
        a1 = np.maximum(0, z1)  # relu: zera o que for negativo
        z2 = W2 @ a1 + b2  # combinacao linear da saida
        a2 = sigmoid(z2)   # sigmoid: converte pra probabilidade entre 0 e 1

        # backprop: volta atualizando os pesos com base no loss
        delta2 = 2*(a2 - y[i]) * a2 * (1 - a2)  # gradiente da saida: MSE * derivada do sigmoid
        delta1 = (W2.T @ delta2) * derivada_relu(z1)  # propaga o erro pela transposta de W2

        # atualiza os pesos
        dw2 = np.outer(delta2, a1)  # produto externo: gradiente de cada peso em W2
        W2 = W2 - Lr * dw2
        b2 = b2 - Lr * delta2  # bias tem gradiente igual ao delta

        dw1 = np.outer(delta1, x)  # mesmo esquema para W1, mas usando x como entrada
        W1 = W1 - Lr * dw1
        b1 = b1 - Lr * delta1

    # printa a loss a cada 100 epocas pra ver se ta convergindo
    loss = np.mean((a2 - y[i])**2)
    if epoca % 100 == 0:
        print(f'epoca {epoca}, loss: {loss:.4f}')
```

```python
# testa a rede nos 4 casos do XOR
for i in range(len(X)):
        x = X[i]
        z1 = W1 @ x + b1
        a1 = np.maximum(0, z1)
        z2 = W2 @ a1 + b2
        a2 = sigmoid(z2)

        predicao = (a2 > 0.5).astype(int)  # se passou de 0.5 considera 1, senao 0
        print(f'{X[i]} -> {predicao[0]} (esperado: {y[i][0]})')
```
